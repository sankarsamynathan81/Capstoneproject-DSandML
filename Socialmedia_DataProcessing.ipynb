```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
```

```python
dataset=pd.read_csv("AI_SocialMedia_Student_Dataset.csv")
dataset
```

```python
# Remove unwanted feature
df=dataset.drop(columns=['Student_ID'])
df
```

```python
# Checking for null or missing value
df.isnull().sum()
```

```python
# Converting categorical data into numeric
df = pd.get_dummies(df, dtype= int, drop_first=True)
df
```

```python
#checking dtypes
df.dtypes
```

```python
# Checking column names
df.columns
```

```python
df.info()
```

```python
#converting all decimal values into numberical values - This will avoid the continous values 
df['Sleep_Hours'] = df['Sleep_Hours'].apply(np.int64)
df['Daily_Social_Media_Hours'] = df['Daily_Social_Media_Hours'].apply(np.int64)
df['Daily_AI_Tool_Usage_Hours'] = df['Daily_AI_Tool_Usage_Hours'].apply(np.int64)
df['Physical_Activity_Hours'] = df['Physical_Activity_Hours'].apply(np.int64)
df['Mental_Health_Score'] = df['Mental_Health_Score'].apply(np.int64)
df['Physical_Health_Score'] = df['Physical_Health_Score'].apply(np.int64)
```

```python
df.info()     
```

```python
# Saving Cleaned Dataset.
df.to_csv("AI_SocialMedia_Student_Dataset_DP.csv",index=False)
```
